## Task1 قراءة الكود قبل التشغيل

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Sara", "Noura", "Reem", "Hind"],
    "city": ["Riyadh", "Jeddah", "Riyadh", "Dammam"],
    "score": [88, 92, 75, 60]
}, index=["a", "b", "c", "d"])

print(type(df["score"]))
print(type(df[["score"]]))
print(df.loc["b", "city"])
print(df.iloc[1, 1])
print(df.loc["a":"c", "score"].tolist())
print(df.iloc[0:2, 2].tolist())
print(df[df["score"] > 80]["name"].tolist())
print(df["city"].value_counts()["Riyadh"])
```

نستخدم الأولى إذا نحتاج عمودًا واحدًا، والثانية إذا نحتاج الاحتفاظ بشكل الجدول.

ولماذا نحتاج Pandas؟**

NumPy array يستخدم للتعامل مع البيانات الرقمية والعمليات الحسابية

##Task3: نظف السجل وسجّل قراراتك

```python
#1. البيانات

data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108, 105, 109, 110, 111],
    "customer": ["Sara", "Noura", "Reem", "Hind", "Lama", "Maha", "Rana", "Dana", "Lama", "Joud", "Hala", "Amal"],
    "city": ["Riyadh", "riyadh ", "Jeddah", "Dammam", "Jeddah", "RIYADH", "Riyadh", None, "Jeddah", "Dammam", "Riyadh", "Jeddah"],
    "drink": ["Latte", "Mocha", "Latte", "Espresso", "Latte", "Mocha", "Latte", "Espresso", "Latte", "Latte", None, "Mocha"],
    "price": ["18", "21", "18", "12", "18", "21", "eighteen", "12", "18", "18", "21", "21"],
    "quantity": [2, 1, 3, 1, 2, None, 1, 4, 2, 1, 2, 1],
    "rating": [5, 4, 5, 3, 4, 5, 4, None, 4, 5, 30, 4] }

df = pd.DataFrame(data)
```

```python
#2. اكتشاف القيم المكتررة والناقصة

print(df.isna().sum())

print(df.duplicated().sum())
```

```python
#3. حذف الصف المتكرر

df = df.drop_duplicates()

print(df)
print("عدد الصفوف بعد الحذف:", len(df))
```

```python
#4. توحيد كتابة المدن

df["city"] = df["city"].str.strip()
df["city"] = df["city"].str.title()

print(df["city"].value_counts())
```

```python
#5. تحويل السعر الى رقم

df["price"] = df["price"].replace("eighteen", "18")

df["price"] = df["price"].astype(float)
print(df["price"])
print(df["price"].dtype)
```

```python
#6. اكتشاف التقييم الشاذ

df.loc[df["rating"] > 5, "rating"] = None
print(df["rating"])
```

```python
# حساب وسيط الكمية

print(df["quantity"].median())
```

```python
#7. تعبئة القيم الناقصة

df["quantity"] = df["quantity"].fillna(df["quantity"].median())
df["rating"] = df["rating"].fillna(df["rating"].median())
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["drink"] = df["drink"].fillna("Unknown")

print(df.isna().sum())
```

كل الأعمدة الآن تظهر 0 وهذا يعني ما بقي عندنا أي قيمة ناقصة.

```python
#8. طباعة جدول نظيف

print(df)
```

##Task5: وكالة السفر عبر الزمن

```python
trips = { "traveler": ["Zara", "Omar", "Lina", "Fahad", "Zara", "Nora", "Yousef", "Salma", "Khalid", "Maya"],
    "departure_year": [2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026],
    "destination_year": ["1450", "3026", "-300", "1999", "1450", "2500", "1000000", "1917", "1450", None],
    "age": [29, 35, 41, 27, 29, 33, 38, -5, 45, 31],
    "price": [12000, 30000, 45000, 8000, 12000, 22000, 999999, 15000, 12000, 9000],
    "status": ["done", "done", "cancelled", "done", "done", "DONE", "done", "done", "Done", "done"] }

df = pd.DataFrame(trips)
```

```python
# EDA

print(df.isna().sum())
print(df.duplicated().sum())
print(df.nunique())
print(df["status"].value_counts())
```

**ملاحظات EDA**


1. القيم الناقصة: توجد قيمة ناقصة واحدة في destination_year.
2. الصفوف المكررة: يوجد صف واحد مكرر، وهو رحلة Zara إلى سنة 1450.
3. القيم المختلفة: عمود departure_year يحتوي على قيمة واحدة مختلفة فقط، وهي 2026.

4. حالة الرحلات:

done: عدد 7

cancelled: عدد 1

DONE: عدد 1

Done: عدد 1

```python
# cleaning

df = df.drop_duplicates()

df["status"] = df["status"].str.lower()

df["destination_year"] = pd.to_numeric(df["destination_year"])

print(df)
print(df.dtypes)
```

```python
df = df.dropna(subset=["destination_year"])

df.loc[df["age"] < 0, "age"] = None

print(df)
```

```python
# departure_yea

df = df.drop(columns=["departure_year"])

print(df)
```

في التحليل؟

```python
# حساب مدة السفر

df["years_traveled"] = abs(df["destination_year"] - 2026)

df["direction"] = "future"
df.loc[df["destination_year"] < 2026, "direction"] = "past"

print(df[["traveler", "destination_year", "years_traveled", "direction"]])
```

```python
print(df["destination_year"].value_counts())
```

```python
print(df.groupby("direction")["price"].mean())
# هذا الكود يحسب متوسط أسعار الرحلات المتجهة للماضي والمستقبل
```

```python
done_trips = df[df["status"] == "done"]

print("Completed trips:", len(done_trips))
print("Completed revenue:", done_trips["price"].sum())
```

```python
# صاحب اطول رحلة

longest_trip = df.loc[df["years_traveled"].idxmax()]

print(longest_trip[["traveler", "destination_year", "years_traveled"]])
```
